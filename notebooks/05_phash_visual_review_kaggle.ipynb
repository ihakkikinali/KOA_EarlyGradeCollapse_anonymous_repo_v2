```python
import numpy as np, pandas as pd, matplotlib, glob, os, random
matplotlib.use("Agg"); import matplotlib.pyplot as plt
from PIL import Image
from pathlib import Path
from skimage.metrics import structural_similarity as ssim
from matplotlib.backends.backend_pdf import PdfPages

cands = [Path(p) for p in glob.glob("/kaggle/input/**/train", recursive=True)]
ROOT = next(p.parent for p in cands if (p.parent / "val").exists() and (p.parent / "test").exists())
print("DATA ROOT:", ROOT)
OUT = Path("/kaggle/working/visual_review"); OUT.mkdir(parents=True, exist_ok=True)

def load(rel, size=256):
    return np.asarray(Image.open(ROOT / rel).convert("L").resize((size, size), Image.BILINEAR), dtype=np.float32) / 255.0
def sim(a, b):
    return ssim(a, b, data_range=1.0), float(np.corrcoef(a.ravel(), b.ravel())[0, 1]), float(np.abs(a - b).mean())

rows = []
for i, (sa, ga, pa, sb, gb, pb, h) in enumerate(PAIRS):
    A, Bm = load(pa), load(pb)
    s, r, mad = sim(A, Bm); sf, rf, _ = sim(A, Bm[:, ::-1])
    rows.append(dict(pair_id=i + 1, split_a=sa, file_a=pa, grade_a=ga, split_b=sb, file_b=pb, grade_b=gb,
                     hamming=h, ssim=s, corr=r, mean_abs_diff=mad, ssim_mirrored=sf, verdict="", notes=""))
cand = pd.DataFrame(rows)

# Reference A: same filename identifier, left knee vs mirrored right knee
allf = sorted(glob.glob(str(ROOT / "*" / "*" / "*.png")))
byid = {}
for f in allf:
    n = Path(f).name; byid.setdefault(n[:7], {})[n[7]] = Path(f).relative_to(ROOT).as_posix()
random.seed(0)
ids = [k for k, v in byid.items() if "L" in v and "R" in v]
same_id = []
for k in random.sample(ids, 223):
    A, Bm = load(byid[k]["L"]), load(byid[k]["R"])[:, ::-1]
    same_id.append(sim(A, Bm)[0])
# Reference B: random cross-split pairs with the same grade
bysg = {}
for f in allf:
    rel = Path(f).relative_to(ROOT).as_posix(); sp, g = rel.split("/")[:2]; bysg.setdefault((sp, g), []).append(rel)
rand = []
for (sa, ga, pa, sb, gb, pb, h) in PAIRS:
    pool_a = bysg.get((sa, str(ga))) or [x for (sp, g), v in bysg.items() if sp == sa for x in v]
    pool_b = bysg.get((sb, str(ga))) or [x for (sp, g), v in bysg.items() if sp == sb for x in v]
    pa2 = random.choice(pool_a); pb2 = random.choice(pool_b)
    rand.append(sim(load(pa2), load(pb2))[0])
ref = pd.DataFrame({"group": ["candidate pairs"] * len(cand) + ["same identifier L vs mirrored R"] * len(same_id) + ["random same-grade cross-split"] * len(rand),
                    "ssim": list(cand.ssim) + same_id + rand})
summary = ref.groupby("group").ssim.describe(percentiles=[0.05, 0.5, 0.95]).round(3)
print(summary); summary.to_csv(OUT / "ssim_reference_summary.csv")
print("Candidate pairs with SSIM > 0.95:", int((cand.ssim > 0.95).sum()), "| > 0.90:", int((cand.ssim > 0.90).sum()))
cand["ssim_rank"] = cand.ssim.rank(ascending=False, method="first").astype(int)
cand.sort_values("pair_id").to_csv(OUT / "visual_review_template.csv", index=False)

fig, ax = plt.subplots(figsize=(6, 3.2))
for g, c in zip(["random same-grade cross-split", "candidate pairs", "same identifier L vs mirrored R"], ["0.6", "#d62728", "#1f77b4"]):
    ax.hist(ref[ref.group == g].ssim, bins=30, alpha=0.55, color=c, label=g)
ax.set_xlabel("SSIM (256 x 256 grey level)"); ax.set_ylabel("pairs"); ax.legend(fontsize=7, frameon=False)
fig.tight_layout(); fig.savefig(OUT / "ssim_reference_distributions.png", dpi=200); plt.close(fig)
```
